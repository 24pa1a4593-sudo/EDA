Handling Inconsistent Values in the data **set**

```python
df['department'] = df['department'].str.strip().str.lower()

department_mapping = {
    'finance': 'Finance',
    'finace': 'Finance',
    'inance': 'Finance',
    'financ': 'Finance',
    'hr': 'HR',
    'h': 'HR',
    'r': 'HR',
    'sales': 'Sales',
    'sals': 'Sales',
    'saes': 'Sales',
    'ales': 'Sales',
    'engineering': 'Engineering',
    'engineerig': 'Engineering',
    'engneering': 'Engineering',
    'support': 'Support',
    'suport': 'Support',
    'suppot': 'Support',
    'upport': 'Support',
    'spport': 'Support'
}

df['department'] = df['department'].map(department_mapping)
```

```python
df['work_accident'] = df['work_accident'].str.strip().str.lower()

df['work_accident'] = df['work_accident'].replace({
    'yes': 'Yes',
    'no': 'No',
    'n': 'No',
    'o': 'No'
})

print(df['work_accident'].value_counts())
```

```python
df['education_level'] = df['education_level'].str.strip().str.lower()

education_mapping = {
    'high school': 'High School',
    'highschool': 'High School',
    'high chool': 'High School',
    'high scool': 'High School',
    'igh school': 'High School',
    'high schol': 'High School',
    'hig school': 'High School',
    'bachelors': 'Bachelors',
    'bacelors': 'Bachelors',
    'achelors': 'Bachelors',
    'bachlors': 'Bachelors',
    'bachelos': 'Bachelors',
    'bahelors': 'Bachelors',
    'bacheors': 'Bachelors',
    'masters': 'Masters',
    'msters': 'Masters',
    'maters': 'Masters',
    'masers': 'Masters',
    'phd': 'PhD',
    'ph': 'PhD',
    'pd': 'PhD',
    'hd': 'PhD'
}

df['education_level'] = df['education_level'].map(education_mapping)

print(df['education_level'].value_counts())
```

Handling Missing Values in the **dataset**

```python
numerical_columns = [
    'salary_k',
    'years_at_company',
    'last_performance_score'
]

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

print(df.isnull().sum())
```

Applying the mode method to fill the missing values **in** categorial data column

```python
categorical_columns = [
    'department',
    'work_accident',
    'education_level',
    'promotion_last_5years',
    'left_company'
]

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

detecting the **outlier**

```python

for col in ['salary_k', 'years_at_company',
            'last_performance_score']:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)
    ]

    print(col, "Outliers:", len(outliers))
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df[
    ['salary_k', 'years_at_company', 'last_performance_score']
])

plt.title("Outlier Detection Using Boxplot")
plt.show()
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=[
        'department',
        'work_accident',
        'education_level',
        'promotion_last_5years'
    ],
    drop_first=True,
    dtype=int
)

print(df_encoded.head())
```
